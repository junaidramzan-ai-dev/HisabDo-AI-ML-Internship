```python
import pandas as pd
```

```python
data = {
    "Name": [
        "Ali", "Ahmed", "Fatima", "Ayesha", "Usman",
        "Hamza", "Hassan", "Zain", "Sara", "Noor",
        "Bilal", "Kashif", "Hina", "Maryam", "Danish",
        "Imran", "Laiba", "Saad", "Areeba", "Talha"
    ],
    "Age": [
        20, 21, 19, 22, 20,
        23, 21, 20, 19, 22,
        20, 21, 20, 19, 23,
        22, 20, 21, 19, 22
    ],
    "Course": [
        "Artificial Intelligence", "Data Science", "Cyber Security", "Artificial Intelligence", "Software Engineering.",
        "Artificial Intelligence", "Data Science", "Cyber Security", "Artificial Intelligence", "Software Engineering.",
        "Artificial Intelligence", "Data Science", "Cyber Security", "Artificial Intelligence", "Software Engineering.",
        "Artificial Intelligence", "Data Science", "Cyber Security", "Artificial Intelligence", "Software Engineering."
    ],
    "Marks": [
        88, 74, 91, 68, 82,
        77, 59, 85, 94, 71,
        80, 66, 89, 73, 64,
        92, 87, 76, 81, 69
    ]
}
```

```python
try:
    df = pd.DataFrame(data)
except Exception as e:
    print(f"Error Creating DataFrame: {e}")
```

```python
print("Hello, World!")
```

```python
df.head()
```

```python
df
```

```python
marks_over_70 = df[df["Marks"] > 70]
print("Marks Greater than 70:")
display(marks_over_70)
```

```python
average_marks = df["Marks"].mean()
display(f"Average of Marks : {average_marks}")
```

```python
highest_marks = df["Marks"].max()
display(f"Highest Marks : {highest_marks}")
```

```python
lowest_marks = df["Marks"].min()
display(f"Lowest Marks : {lowest_marks}")
```
